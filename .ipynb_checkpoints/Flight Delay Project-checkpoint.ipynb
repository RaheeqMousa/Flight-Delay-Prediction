# Sampling

create a new column weather the flight is delayed or not (0 = not delayed / 1 = delayed)

```python
df['delayed']=(df['arr_delay']>=15).astype(int)
```

Making Stratified Sampling while keepingg aware of the distribution stay the same

```python
print(f"Before a sampling, ratio of delayed flights {df['delayed'].mean()} and ratio of non delayed {(df['delayed']==0).mean()}")

total=500000
delayed_ratio= df['delayed'].mean()

number_delayed= int(total*delayed_ratio)
number_non_delayed= total-number_delayed

delayed_flights=df[df['delayed']==1].sample(n=number_delayed,random_state=7)
non_delayed_flights=df[df['delayed']==0].sample(n=number_non_delayed,random_state=7)

sampled_flights=pd.concat([delayed_flights,non_delayed_flights]).sample(frac=1, random_state=4)
print(f"After a stratified sampling, ratio of delayed flights {number_delayed/total} and ratio of non delayed {number_non_delayed/total}")
```

Write the sampled data to a new file

```python
os.makedirs('data/sampled', exist_ok=True)
```

```python
sampled_flights.to_csv('data/sampled/sampled_flights.csv')
```

## 2. Solving Inconsistency

Here we are checking the string columns, if they have an entry containins additional space (not trimmed string)

```python
str_cols = ['origin', 'origin_city_name', 'origin_state_nm',
            'dest', 'dest_city_name', 'dest_state_nm',
            'op_unique_carrier', 'op_carrier_fl_num']

for c in str_cols:
    raw   = df[c].nunique()
    non_trimmed_strings = df[c].str.strip().str.upper().nunique()
    if raw == non_trimmed_strings:
        print(f"no issues at {c} ({raw} unique values)")
    else:
        print(f"{c}: {raw - non_trimmed_strings} values are not trimmed")
```

```python
str_cols = ['origin', 'origin_city_name', 'origin_state_nm',
            'dest', 'dest_city_name', 'dest_state_nm',
            'op_unique_carrier', 'op_carrier_fl_num']
for c in str_cols:
    raw   = df[c].nunique()
    capital_strings = df[c].str.upper().nunique() #this convert the string to capital then counts unique values
    if raw == capital_strings:
        print(f"No case problems at {c}, and ({raw} unique values)")
    else:
        print(f"There is a case problem at {c}, and ({raw-capital_strings} have problems)")
```

## 5. Outliers

Here we created a column called distance_zscore where we measured how many standard deviation each value is from the mean in standard deviations.<br> 
Then we created a column called distance_outlier that marks values as outlier if greater or equal to 3.<br>
As shown, the result display the number of outlier and non-outlier observaions in the dataset.

```python
df['distance_zscore']= (df['distance']-df['distance'].mean())/df['distance'].std()
df['distance_outlier'] = df['distance_zscore'].apply(lambda x: abs(x) >= 3)
df["distance_outlier"].value_counts()
```

```python
(df["distance_outlier"]==True).sum()/len(df["distance_outlier"])
```

# Add features

Add a season column for the flights<br>
it assigns a season label to each flight based on its month.

```python
seasons = []

for m in df['month']:
    if m <= 2:
        seasons.append('winter')
    elif m <= 5:
        seasons.append('spring')
    elif m <= 8:
        seasons.append('summer')
    else:
        seasons.append('fall')

df['season'] = seasons
```

We introduced a new feature called “is_weekend” <br>which assigns a binary value to each flight indicating whether it occurs in weekend or not (0=not weekend, 1=weekend). 

```python
df['is_weekend'] = df['day_of_week'] >= 6
```

The feature type of "is_weekend" now is bool and we want to encode it into 0/1 (0=not weekend, 1=weekend).

```python
df.info()
```

We converted the type of Here we converted the is_weekend feature into and integer type

```python
df['is_weekend'] = df['is_weekend'].astype(int)
```

As you see is_weekend feature became an integer type. 

```python
df.info()
```

We converted the type of distance_outlier from boolean to integer (1=is distance outlier, 0= not a distance outlier). 

```python
df['distance_outlier'] = df['distance_outlier'].astype(int)
```

As shown below, the datatype of distance_outlier now is an integer. 

```python
df.info()
```

We first extracted the departure hour from the scheduled departure time (crs_dep_time).<br>
Then, we created a new feature called period_of_day, which categorizes each flight as morning, afternoon, evening, or night based on its departure hour.

```python
df['departure_hour'] = df['crs_dep_time'] // 100
```

```python
result = []

for h in df['departure_hour']:
    if 5 <= h <= 11:
        result.append('morning')
    elif 12 <= h <= 16:
        result.append('afternoon')
    elif 17 <= h <= 20:
        result.append('evening')
    else:
        result.append('night')

df['period_of_day'] = result
```

As you see in the image below, the periodofday feature is added successfully.

# Correlation Heatmap

Need to see the features that can affect the delay column and the features that they don't have a strong effect on it

```python
numeric_cols = [
    'month',
    'day_of_month',
    'day_of_week',
    'distance',
    'origin_lat',
    'origin_lng',
    'dest_lat',
    'dest_lng',
    'departure_hour',
    'delayed'
]

corr = df[numeric_cols].corr()

plt.figure(figsize=(12,8))

sns.heatmap(
    corr,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Heatmap')

plt.show()
```

Depends on the heat map most features have a weak linear correlation with delay field, that means the flight delay are influenced by many factors (features) not only one factor.

But as what i say before that the time of flight delay in day affect the flights after it so thats whay it is stronger a bit than the others

**To prove what i said we will check the delay perventage throw the day**

```python
hour_delay = df.groupby('departure_hour')['delayed'].mean() * 100

plt.figure(figsize=(14,6))
sns.lineplot(x=hour_delay.index, y=hour_delay.values, marker='o', color='crimson')
plt.axhline(y=df['delayed'].mean()*100, color='gray', linestyle='--', label='Average Delay Rate')
plt.fill_between(hour_delay.index, hour_delay.values, alpha=0.2, color='red')
plt.title('How Delay Risk Grows Throughout the Day')
plt.xlabel('Hour of Day (0 = Midnight)')
plt.ylabel('Delay Percentage (%)')
plt.legend()
plt.show()
```

As we see the lowest delay risk is on 5-6 Am and after that the delay risk growth, and when we reach 14 pm the delay exceed the average delay percentage to reach at max the 30% chance to delay

<p style="font-weight:bold">Define the target vecotr and the feature matrix.</p>

```python
df_model = df_model.drop(columns=['dest_city_name','origin_city_name','fl_date','op_carrier_fl_num','Unnamed: 0','distance_zscore','distance_outlier'])

# cyclical encoding of departure time (take the sin and cos of the hour of the day)
df_model['dep_hour_sin'] = np.sin(2 * np.pi * df_model['departure_hour'] / 24)
df_model['dep_hour_cos'] = np.cos(2 * np.pi * df_model['departure_hour'] / 24)

#   route-level historical delay rate 
# 'What % of flights on this exact route are delayed?' is very predictive
route_delay = df_model.groupby(['origin', 'dest'])['delayed'].mean()
df_model['route_delay_rate'] = df_model[['origin', 'dest']].apply(
    lambda r: route_delay.get((r['origin'], r['dest']), 0), axis=1
).values

#   carrier historical delay rate 
carrier_delay = df_model.groupby('op_unique_carrier')['delayed'].mean()
df_model['carrier_delay_rate'] = df_model['op_unique_carrier'].map(carrier_delay)

#  month cyclical encoding
df_model['month_sin'] = np.sin(2 * np.pi * df_model['month'] / 12)
df_model['month_cos'] = np.cos(2 * np.pi * df_model['month'] / 12)


y= df_model['delayed'] #target vecor
x= df_model.drop(columns=['delayed']) #feature matrix
```

```python
from sklearn.model_selection import validation_curve
from sklearn.tree import DecisionTreeClassifier

param_range = np.arange(1, 16)  # Tests depths from 1 to 15

# 2. Automatically test all options using 5-Fold Cross-Validation
train_scores, test_scores = validation_curve(
    make_pipeline(preprocessor, DecisionTreeClassifier(class_weight='balanced', random_state=42)),
    x_train, y_train, 
    param_name="decisiontreeclassifier__max_depth", 
    param_range=param_range,
    cv=5, 
    scoring="roc_auc", # Evaluates based on ROC-AUC score
    n_jobs=-1,
    error_score="raise"
)

# 3. Calculate the average scores for training and testing
mean_train_scores = np.mean(train_scores, axis=1)
mean_test_scores = np.mean(test_scores, axis=1)

# 4. Find the mathematically best setting
best_index = np.argmax(mean_test_scores)
best_depth = param_range[best_index]

# 5. Plot the chart to visualize the results
plt.plot(param_range, mean_train_scores, label="Training Score", color="darkorange", lw=2, marker='o')
plt.plot(param_range, mean_test_scores, label="Testing (CV) Score", color="navy", lw=2, marker='s')

# Add a vertical line to mark the absolute best option
plt.axvline(x=best_depth, color="red", linestyle="--", label=f"Best Depth ({best_depth})")

plt.title("Finding the Best max_depth (Validation Curve)")
plt.xlabel("Tree Depth (max_depth)")
plt.ylabel("ROC-AUC Score")
plt.grid(True, linestyle=":", alpha=0.6)
plt.legend(loc="best")
plt.tight_layout()

# Save the chart
plt.savefig("validation_curve.png", dpi=150)
plt.show()
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import roc_auc_score

# Build the baseline decision tree and evaluate it on the test set
dt_base = DecisionTreeClassifier(
    max_depth=7,
    min_samples_leaf=50,
    class_weight='balanced',
    random_state=42
)

dt_pipe = make_pipeline(preprocessor, dt_base)
dt_pipe.fit(x_train, y_train)
y_pred_base = dt_pipe.predict(x_test)

print('=== Baseline Decision Tree ===')
print(f'Accuracy : {accuracy_score(y_test, y_pred_base):.4f}')
print(f'ROC-AUC  : {roc_auc_score(y_test, dt_pipe.predict_proba(x_test)[:,1]):.4f}')
print()
print(classification_report(y_test, y_pred_base, target_names=['On-time', 'Delayed']))
```

## 3. Hyperparameter Tuning 

 this code is finding the best settings for a machine learning model to get  accurate result . take only 30% form the datast to make it faster

```python
from sklearn.model_selection import GridSearchCV

# Keep this cell self-contained even if earlier setup cells were not run.
RANDOM_STATE = 42

# Sample for faster grid search
SAMPLE_FRAC = 0.3
X_tr_s = x_train.sample(frac=SAMPLE_FRAC, random_state=42)
y_tr_s = y_train.loc[X_tr_s.index]

param_grid = {
    'decisiontreeclassifier__max_depth'        : [6, 8, 10, 12, 15],
    'decisiontreeclassifier__min_samples_leaf' : [20, 50, 100],
    'decisiontreeclassifier__min_samples_split': [2, 20, 50],
    'decisiontreeclassifier__criterion'        : ['gini', 'entropy'],
    'decisiontreeclassifier__max_features'     : [None, 'sqrt', 0.7],  # limit features per split — reduces overfitting
}

grid_search = GridSearchCV(
    make_pipeline(preprocessor, DecisionTreeClassifier(class_weight='balanced', random_state=RANDOM_STATE)),
    param_grid,
    cv=3,
    scoring='roc_auc',
    n_jobs=-1,
    verbose=1
)

grid_search.fit(X_tr_s, y_tr_s)
print('\nBest params :', grid_search.best_params_)
print('Best CV AUC :', f"{grid_search.best_score_:.4f}")
```

### Interpretation of Hyperparameter Tuning 
The **ROC-AUC score of 0.67** be the result we can see that the model have a mid ability to find if the flight going to delayed or not

## 6. Optimal Decision Threshold

```python
from sklearn.metrics import precision_recall_curve, classification_report
from sklearn.tree import DecisionTreeClassifier

if 'y_proba' not in globals():
    if 'dt_best' in globals():
        y_proba = dt_best.predict_proba(x_test)[:, 1]
    else:
        dt_base = DecisionTreeClassifier(
            max_depth=7,
            min_samples_leaf=50,
            class_weight='balanced',
            random_state=42
        )
        dt_pipe = make_pipeline(preprocessor, dt_base)
        dt_pipe.fit(x_train, y_train)
        y_proba = dt_pipe.predict_proba(x_test)[:, 1]

# Default threshold is 0.5 — tuning it improves precision/recall tradeoff
precision_vals, recall_vals, thresholds = precision_recall_curve(y_test, y_proba)

# Find threshold that maximises F1 for the Delayed class
f1_scores = 2 * (precision_vals * recall_vals) / (precision_vals + recall_vals + 1e-9)
best_idx = np.argmax(f1_scores)
best_threshold = thresholds[best_idx]
print(f'Best threshold : {best_threshold:.3f}  (default was 0.500)')
print(f'Best F1        : {f1_scores[best_idx]:.4f}')

y_pred_tuned = (y_proba >= best_threshold).astype(int)
print()
print('=== Results with tuned threshold ===')
print(classification_report(y_test, y_pred_tuned, target_names=['On-time', 'Delayed']))

# Plot Precision / Recall / F1 vs threshold
fig, ax = plt.subplots(figsize=(6, 4))
ax.plot(thresholds, precision_vals[:-1], label='Precision', color='steelblue')
ax.plot(thresholds, recall_vals[:-1], label='Recall', color='tomato')
ax.plot(
    thresholds,
    2 * (precision_vals[:-1] * recall_vals[:-1]) / (precision_vals[:-1] + recall_vals[:-1] + 1e-9),
    label='F1', color='green', linestyle='--'
)
ax.axvline(best_threshold, color='black', linestyle=':', label=f'Best threshold ({best_threshold:.2f})')
ax.set_xlabel('Threshold')
ax.set_title('Precision / Recall / F1 vs Threshold (Delayed class)')
ax.legend()
plt.tight_layout()
plt.show()
```

### Interpretation 
we can see from the chart that 0.5 is the best threshold for the model to get great result without decrease F1 , Recall , precison persantage

### OVERALL Interpretation For Decision Tree Model

our Decision Tree model has mid result which it has  highly reliable for detect "On-time" flights, but it struggles significantly with false alarms when predicting delays.

and this is happen because the dataset is heavily imbalanced (only ~21% of flights are actually delayed), so i try to trained with class_weight=balanced. This forced it to take  delays. after that the model successfully catches 64% of all actual delays (Recall), but at a steep cost: when it flags a flight as "Delayed," it is only correct 32% of the time (Precision).


The model relies heavily on historical trends and timing. route_delay_rate (33.5%) and crs_dep_time (31.5%) combine to drive over 65% of the model's decisions. Simply put, where the flight is going (and its route's history) and what time of day it leaves are the ultimate deciders of a delay.

The Threshold is Optimized: i chosen threshold of 0.508 which it the best possible cutoff for this  model to maximize the F1-Score and we can see that on the Optimal Decision Threshold chart. Moving it higher would stop the false alarms but cause the model to miss almost all real delays.

<p style="font-weight:bold;">Random Forest parameters with best threshold, exp.5</p>

we want to choose the threshold that gives the highest F1-score.<br>
F1-score balances precision and recall, so we want to select threshold that provides the best trade-off between them.

```python
from sklearn.model_selection import cross_val_predict

y_proba = cross_val_predict(
    rf_pipeline,
    x_train,
    y_train,
    cv=5, #cross validation, data here is splitted into 5 folds. In each iteration, 4 folds for training and 1 for validation
    method="predict_proba" #this returns the probability of predicting classes instead of class labels
)[:, 1] #returns the probability of class 1(delayed class)
```

```python
from sklearn.metrics import precision_recall_curve

precision, recall, thresholds = precision_recall_curve(y_train, y_proba)

f1_scores = 2 * (precision[:-1] * recall[:-1]) / (precision[:-1] + recall[:-1] + 1e-9) #1e-9 prevents dividing by zero if recall and precision were both 0

best_idx = np.argmax(f1_scores)#the index of the maximum value
best_threshold_pr = thresholds[best_idx] 

print("Best threshold (Precision/Recall):", best_threshold_pr)
```

The threshold we found is approximatley equal the default threshold (0.5), meaning that predicting on that threshold won't add benifit to the results before.

### 3. XGBoost

XGBoost (Extreme Gradient Boosting) builds decision trees sequentially: each new tree is trained to correct the errors of the trees before it, and the final output prediction is the sum of all the trees. This is different from Random Forest, which builds many independent trees and averages them. We train it on the same `preprocessor`, `x_train/x_test/y_train/y_test` so we can compare it with Random Forest above, and we follow the same evaluation order (confusion matrix -> accuracy -> precision -> recall -> f1 -> classification report).

```python
pip install xgboost
```

```python
from xgboost import XGBClassifier
```

```python
xgb= XGBClassifier(
    n_jobs=-1,
    eval_metric="logloss",
    random_state=74,
)

xgb_pipeline=make_pipeline(preprocessor, xgb)

xgb_pipeline.fit(x_train,y_train)
xgb_predict= xgb_pipeline.predict(x_test)
```

```python
ConfusionMatrixDisplay.from_predictions(y_test,xgb_predict)
```

```python
#accuracy on train set
xgb_predict_train= xgb_pipeline.predict(x_train)
accuracy_score(y_train, xgb_predict_train)
```

```python
#accuracy on test set
accuracy_score(y_test, xgb_predict)
```

Accuracy in training phase achieved approximatley 0.82, while in test set it achieved 0.79.<br>
The two scores are close, so it is different than the Random Forest the model is not strongly overfitting. However this high accuracy is not correct 100% because the data is imbalanced (about 79% non-delayed), so a model that mostly predicts "non-delayed" already looks good.

<p style="font-weight:bold;">Applying class balancing (scale_pos_weight) to XGBoost</p>

scale_pos_weight is the XGBoost equivalent of class_weight="balanced" <br>
it tells the model how much more to weight the positive (delayed) class.

```python
scale_pos_weight = (y_train == 0).sum() / (y_train == 1).sum()

xgb= XGBClassifier(
    n_jobs=-1,
    scale_pos_weight=scale_pos_weight,
    eval_metric="logloss",
    random_state=74,
)

xgb_pipeline=make_pipeline(preprocessor, xgb)
xgb_pipeline.fit(x_train,y_train)
xgb_predict= xgb_pipeline.predict(x_test)
```

```python
xgb_predict_train= xgb_pipeline.predict(x_train)
print(accuracy_score(y_train, xgb_predict_train))
```

```python
print(accuracy_score(y_test, xgb_predict))
```

Accuracy in training phase achieved 0.73, while in test set it achieved 0.67.<br>
The accuracy dropped compared to the baseline, but this is expected: with scale_pos_weight the model now pays attention to the minority (delayed) class instead of ignoring it.

```python
print(precision_score(y_test, xgb_predict))
```

between all flights that are classified as positive(delayed). about 33% of them were correct

```python
print(recall_score(y_test, xgb_predict))
print(f1_score(y_test, xgb_predict))
```

out of all positive cases (delayed flights) the model now found about 59% of them (up from 10%)<br>
and the f1-score for the delayed class rose to 0.43

```python
print(classification_report(y_test, xgb_predict))
```

On class=0 (non delayed flights) the model achieved precision=0.87, recall=0.69, and f1-score=0.77.<br>
on class=1 (delayed flights) the model achieved precision=0.34, recall=0.60, and f1-score=0.43, so it now catches 60% of delayed flights instead of 10%, but raises more false alarms which lowers precision.<br>
accuracy dropped to 0.67 because the model no longer just predicts the majority class.<br>
This is the same trade-off we saw with the Random Forest after class_weight="balanced".

The behaviour matches the Random Forest: scale_pos_weight raises recall for the delayed class at the cost of precision and overall accuracy. The weak correlation between the features and the "delayed" target (shown in the correlation analysis above) limits how much any model can improve.

<p style="font-weight:bold;">XGBoost parameters tuning after class balancing exp.1</p>

```python
xgb = XGBClassifier(
    n_estimators=300,
    max_depth=6,
    learning_rate=0.1,
    subsample=0.8,
    colsample_bytree=0.8,
    scale_pos_weight=scale_pos_weight,
    eval_metric="logloss",
    n_jobs=-1,
    random_state=74,
)
xgb_pipeline=make_pipeline(preprocessor, xgb)

xgb_pipeline.fit(x_train,y_train)
xgb_predict= xgb_pipeline.predict(x_test)
```

max_depth=6 keeps each tree shallow so boosting combines many weak learners.<br>
learning_rate=0.1 controls how much each new tree contributes (smaller = slower but more robust).<br>
n_estimators=300 is the number of boosting rounds (trees built one after another).<br>
subsample=0.8 and colsample_bytree=0.8 sample 80% of the rows/features per tree to fight overfitting.

```python
xgb_predict_train= xgb_pipeline.predict(x_train)
print(accuracy_score(y_train, xgb_predict_train))
```

```python
print(accuracy_score(y_test, xgb_predict))
```

The model achieved 0.75 on train and 0.67 on test. The gap become small, so the model is not overfitting, but the accuracy is almost identical to the simple balanced model above.

```python
print(precision_score(y_test, xgb_predict))
```

between all flights that are classified as positive(delayed). about 33% of them were correct

```python
print(recall_score(y_test, xgb_predict))
print(f1_score(y_test, xgb_predict))
```

out of all delayed flights the model found about 58% of them, with an f1-score of 0.42 for the delayed class

```python
print(classification_report(y_test, xgb_predict))
```

Tuning the depth, learning rate and number of trees barely changed the result (class 1 precision=0.34, recall=0.58, f1=0.43). The model behaves almost the same as the default balanced XGBoost.

<p style="font-weight:bold;">XGBoost parameters tuning after class balancing exp.2</p>
